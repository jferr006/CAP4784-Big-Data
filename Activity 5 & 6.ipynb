```python
from pyspark.sql import functions
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('data_processing'). getOrCreate()
```

```python
import pyspark.sql.functions as F
```

```python
from pyspark.sql.types import *
```

```python
schema=StructType().add("user_id","string").add("country","string").add("browser","string").add("OS",'string').add("age", "integer")
```

```python
df=spark.createDataFrame([("A203",'India',"Chrome","WIN",
33),("A201",'China',"Safari","MacOS",35),("A205",'UK',"Mozilla",
"Linux",25),("A207",'Greenland',"Chrme","WIND",
29)
],schema=schema)
```

```python
df.printSchema()
```

```python
df.show()
```

```python
df_na=spark.createDataFrame([("A203",None,"Chrome","WIN",
33),("A201",'China',None,"MacOS",35),("A205",'UK',"Mozilla",
"Linux",25), ("A207",'Greenland',"Chrme",None,
29)
],schema=schema)
```

```python
df_na.show()
```

```python
df_na.fillna('0').show()
```

```python
df_na.fillna( { 'country':'USA', 'browser':'Safari','OS':'WIN' } ).show()
```

```python
df_na.na.drop().show()
```

```python
df_na.na.drop(subset='country').show()
```

```python
df_na.replace("Chrome","Google Chrome").show()
```

```python
df_na.drop('user_id').show()
```

```python
"6 Create Schema"
schema=StructType().add("officer_no","string").add("vehicle_no","string").add("current_rank","string").add("gender",'string').add("years_of_service", "integer")
```

```python
"7 Create Dataframe"
Officer_Data = spark.createDataFrame([
    ("OFF-101", "VEH-501", "Officer", "Female", 3),
    ("OFF-102", "VEH-502", "Sergeant", "Male", 12),
    ("OFF-103", "VEH-503", "Lieutenant", "Female", 18),
    ("OFF-104", "VEH-504", "Captain", "Male", 22),
    ("OFF-105", "VEH-505", "Officer", "Male", 5)
], schema=schema)
```

```python
"8 Display"
display(Officer_Data)
```

```python
"9 Display/show"
Officer_Data.show()
```

```python
"10 new schema with nulls"
Officer_Data = spark.createDataFrame([
    ("OFF-101", "VEH-501", "Officer", "Female", 3),
    ("OFF-102", None, "Sergeant", "Male", 12),
    ("OFF-103", "VEH-503", None, "Female", 18),
    ("OFF-104", "VEH-504", "Captain", None, 22),
    ("OFF-105", "VEH-505", "Officer", "Male", None)
], schema=schema)
```

```python
"11 Print schema"
Officer_Data.printSchema()
```

```python
"12 Replace"
Officer_Data.fillna('0').show()
```

```python
"13 Fill Nulls"
Officer_Data.fillna( { 'vehicle_no':'SERVICE', 'current_rank':'PENDING','gender':'N/A', 'years_of_service':'0' } ).show()
```

```python
"14 Drop"
Officer_Data.na.drop().show()
```

```python
"15 Filter"
Officer_Data.na.drop(subset='gender').show()
```

```python
"16 Replace"
Officer_Data.replace("Captain","Station Cheif").show()
```

```python
"17 Drop"
Officer_Data.drop('vehicle_no').show()
```

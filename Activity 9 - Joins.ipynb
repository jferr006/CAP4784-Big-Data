```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('data_processing_tasks'). getOrCreate()
```

```python
# 5- Run  the code below. Explain the syntax. 

import pyspark.sql.functions as F

# 6- Run the code below. Explain the syntax.

from pyspark.sql.types import *

# Syntax explained: Loads PySpark’s built-in functions (like summing numbers or formatting text) and gives them the shortcut name F so they are faster to type. / Imports all PySpark data types (like StringType for text or IntegerType for whole numbers) so you can build table schemas directly.
```

```python
# 7- Run the code below. Explain the syntax and list your observations. 

region_data = spark.createDataFrame([('Family with grownups','PN'),('Driven Growers','GJ'),('Conservative families','DD'),('Cruising Seniors','DL'),
('Average Family ','MN'),('Living well','KA'),('Successful hedonists','JH'),('Retired and Religious','AX'),('Career Loners','HY'),('Farmers','JH')],
schema=StructType().add("Customer_main_type","string").add("Region Code","string"))

# Syntax explained: Creating a dataframe named region_data then using a schema to create two columns.
```

```python
# 8- Run the code below. Explain the syntax and the output. 

region_data.display()

# Syntax explained: Using the display function to display the newly created dataframe.
```

```python
# 9-Run the code below. Explain the syntax. List your observations.

customer=spark.createDataFrame([('Family with grownups',72000),('Driven Growers',19000),('Conservative families',59500),
('Cruising Seniors',29500),('Average Family ',69438),('Living well',92000),('Successful hedonists',28500),('Retired and Religious',25750),
('Career Loners',32900),('Farmers',73000)],schema=StructType().add("Customer_main_type","string").add("Avg_Salary","integer"))

# Syntax explained: We create a new dataframe called customer with a schema adding two columns.
```

```python
# 10- Run the code below. Explain the syntax and the output.

customer.display()

# 11-Change the visualization you obtained above to be a bar graph. List your observations.

# Syntax explained: Use the display function to show the customer dataframe.
# we then make a bar chart of the customer dataframe.
```

```python
3# 12- Run the code below. Explain the syntax.

new_df=customer.join(region_data,on='Customer_main_type')

# Syntax explained: we create a join between the customer and region_data dataframes.
```

```python
# 13-Run the code below.Explain the syntax and the output. List your observations.

new_df.display() 

# 14- Change the visualization you obtained above to be a pie chart showing the average salary by customer_main_type. 

# Syntax explained: use the display function to show the new dataframe of the join between customer and region_data. then we create a pie chart of the new dataframe.
```

```python
# 15- Run the code below and list your observations. Explain the syntax and the output.

new_df.groupby("Region Code").count().show()

# Syntax explained: a new group by is made on the new_df dataframe. which displays the count of each region code.
```

```python
# 15- Run the code below. Explain the syntax and the output.


customer.groupBy('Customer_main_type').sum('Avg_Salary').fillna(0).show()

# Syntax explained: a new group by function is made to the customer dataframe. showing the sum of the average salary. filling any missing values with zero.
# No fields missing.
```

```python
# 16-Run the code below and list your observations. Explain the syntax and the output.

customer.groupBy('Customer_main_type').sum('Avg_Salary').fillna(0).orderBy('sum(Avg_Salary)').show()

# Syntax explained: Ordered by average salary. change nulls to zero. the default is ascending order. the customer with the highest salary is Living Well.
```

```python
# 17-Run the code below and list your observations. Explain the syntax and the output.

customer.groupBy('Customer_main_type').sum('Avg_Salary').fillna(0).orderBy('sum(Avg_Salary)',ascending=False).show()

# Syntax explained: Same information in descending order.
```

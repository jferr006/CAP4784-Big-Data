```python
#Creating the Schema

from pyspark.sql.types import StructType, StructField, StringType, IntegerType

schema = StructType([
    StructField("facility_id", StringType(), False),      # No null values
    StructField("facility_name", StringType(), True),
    StructField("campus", StringType(), True),
    StructField("capacity", IntegerType(), True),
    StructField("description", StringType(), True)
])
```

```python
# Creating the dataframe

data = [
    ("1001", "basketball court", "Kendall", 2500, "inspection passed"),
    ("1002", "baseball field", "Kendall", 600, "fence to be inspected"),
    ("1003", "library", "Hialeah", 240, None),
    ("1004", "Gallery", "North", 120, "Open to the public"),
    ("1005", "Idea Center", None, 200, "Events scheduled")
]

facilities_df = spark.createDataFrame(data, schema=schema)
```

```python
# Display the schema

facilities_df.printSchema()
```

```python
# Display the dataframe

display(facilities_df)
```

```python
#Show the dataframe if we replaced Null values with 0. Would that make sense? Explain
```

```python
# Replacing Null values with '0'
df_filled_zero = facilities_df.fillna('0')

# Display the result
display(df_filled_zero)
```

```python
#Show the dataframe if we filled the null values for the fields campus with "District" and description with "Not Available".
```

```python
# Fill null values for campus with "District" and null values for description with "Not Available"
df_filled_custom = facilities_df.fillna({
    'campus': 'District',
    'description': 'Not Available'
})

# Display the results
display(df_filled_custom)
```

```python
#Show how the existing dataframe would look like with rows including null values being dropped
```

```python
# Drop rows that contain at least one null value
df_dropped_rows = facilities_df.na.drop()

# Display the results
display(df_dropped_rows)
```

```python
#Show how the existing dataframe would look like with rows including null values for 'Description' being dropped
```

```python
# Drop rows where the 'description' column is null
df_dropped_description = facilities_df.na.drop(subset=['description'])

# Display the results
display(df_dropped_description)
```

```python
#Show how the existing dataframe would look like with rows including null values for campus being dropped
```

```python
# Drop rows where the 'campus' column is null
df_dropped_campus = facilities_df.na.drop(subset=['campus'])

# Display the results
display(df_dropped_campus)
```

```python
#Show how the existing dataframe would look like if we replaced "baseball field" with baseball
```

```python
# Replace "baseball field" with "baseball" in the DataFrame
df_replaced_baseball = facilities_df.replace("baseball field", "baseball")

# Display the results
display(df_replaced_baseball)
```
